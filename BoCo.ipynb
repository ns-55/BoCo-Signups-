```python
import pandas as pd #loads the pandas library and names it pd for easy access later
#pandas is the tool used to work with tabular (excel format bu editable) data
df = pd.read_csv("BoCo_200.csv") #reads csv file and creates a dataframe
df.head() #shows first 5 rows of data set - useful to make sure data is in the right format

df['event_date'] = pd.to_datetime(df['event_date']) #to format the date correctly for easier data retreival

df['attended'] = df['attended'].fillna("N-S") #replaces any empty entries in attendance status with the correct entry

df['ref_name'] = df['ref_name'].str.title() #that way if someone enters "Name" and "name" it is viewed as the same entry

df['day_of_week'] = df['event_date'].dt.day_name() #names the days of the week - helpful to identify trends

df['time_in'] = pd.to_datetime(df['time_in'], format="%I:%M %p", errors='coerce')
df['time_out'] = pd.to_datetime(df['time_out'], format="%I:%M %p", errors='coerce')

df['volunteer_hours'] = (
    pd.to_datetime(df['time_out'], errors='coerce') - pd.to_datetime(df['time_in'], errors='coerce')
).dt.total_seconds() / 3600 #time volunteered in hours

import sqlite3

conn = sqlite3.connect("volunteers.db")
df.to_sql("volunteer_activity", conn, if_exists="replace", index=False)
conn.close() #loads SQLite into python; allows me to create a small SQL database store in a file
```
